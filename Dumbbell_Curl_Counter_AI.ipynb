```python
#Extract Shoulder, Elbow, Wrist

import numpy as np

pts = r.keypoints.xy[0].cpu().numpy()

right_shoulder = pts[6]
right_elbow    = pts[8]
right_wrist    = pts[10]

print("Shoulder:", right_shoulder)
print("Elbow:", right_elbow)
print("Wrist:", right_wrist)
```

```python
#Angle Function

import numpy as np

def calculate_angle(a, b, c):

    a = np.array(a)
    b = np.array(b)
    c = np.array(c)

    radians = np.arctan2(
        c[1]-b[1],
        c[0]-b[0]
    ) - np.arctan2(
        a[1]-b[1],
        a[0]-b[0]
    )

    angle = np.abs(radians * 180 / np.pi)

    if angle > 180:
        angle = 360 - angle

    return angle
```

```python
#Calculate Elbow Angle

angle = calculate_angle(
    right_shoulder,
    right_elbow,
    right_wrist
)

print("Elbow Angle =", angle)
```

Elbow Angle = 20.27°

This means the arm is fully bent, exactly what we expect at the top of a dumbbell curl.

160° - 180°  → Arm straight

20° - 40°    → Arm curled

So your angle calculation is working correctly

```python
#Create Angle Function

import numpy as np

def calculate_angle(a, b, c):
    a = np.array(a)
    b = np.array(b)
    c = np.array(c)

    radians = np.arctan2(c[1]-b[1], c[0]-b[0]) - \
              np.arctan2(a[1]-b[1], a[0]-b[0])

    angle = np.abs(radians * 180.0 / np.pi)

    if angle > 180:
        angle = 360 - angle

    return angle
```

```python
#Process Entire Video

import cv2
from ultralytics import YOLO

model = YOLO("yolo11n-pose.pt")

video_path = "7187424-hd_1920_1080_24fps.mp4"

cap = cv2.VideoCapture(video_path)

counter = 0
stage = None

frame_no = 0

while True:

    ret, frame = cap.read()

    if not ret:
        break

    frame_no += 1

    # Process every 3rd frame (faster)
    if frame_no % 3 != 0:
        continue

    results = model(frame, verbose=False)

    if len(results) == 0:
        continue

    if results[0].keypoints is None:
        continue

    pts = results[0].keypoints.xy

    if len(pts) == 0:
        continue

    pts = pts[0].cpu().numpy()

    right_shoulder = pts[6]
    right_elbow = pts[8]
    right_wrist = pts[10]

    angle = calculate_angle(
        right_shoulder,
        right_elbow,
        right_wrist
    )

    # Curl logic
    if angle > 150:
        stage = "down"

    if angle < 50 and stage == "down":
        stage = "up"
        counter += 1

        print("Rep:", counter)

cap.release()

print("\nTotal Repetitions =", counter)
```

```python
#Create Gradio Web App

import gradio as gr
import cv2
import numpy as np
from ultralytics import YOLO

model = YOLO("yolo11n-pose.pt")

def calculate_angle(a, b, c):

    a = np.array(a)
    b = np.array(b)
    c = np.array(c)

    radians = np.arctan2(c[1]-b[1], c[0]-b[0]) - \
              np.arctan2(a[1]-b[1], a[0]-b[0])

    angle = np.abs(radians * 180 / np.pi)

    if angle > 180:
        angle = 360 - angle

    return angle


def count_reps(video_path):

    cap = cv2.VideoCapture(video_path)

    counter = 0
    stage = None

    frame_no = 0

    while True:

        ret, frame = cap.read()

        if not ret:
            break

        frame_no += 1

        if frame_no % 3 != 0:
            continue

        results = model(frame, verbose=False)

        if len(results) == 0:
            continue

        if results[0].keypoints is None:
            continue

        pts = results[0].keypoints.xy

        if len(pts) == 0:
            continue

        pts = pts[0].cpu().numpy()

        shoulder = pts[6]
        elbow = pts[8]
        wrist = pts[10]

        angle = calculate_angle(
            shoulder,
            elbow,
            wrist
        )

        if angle > 150:
            stage = "down"

        if angle < 50 and stage == "down":
            stage = "up"
            counter += 1

    cap.release()

    return f"Total Repetitions: {counter}"
```

```python
#Launch Gradio

demo = gr.Interface(
    fn=count_reps,
    inputs=gr.Video(),
    outputs="text",
    title="Dumbbell Curl Counter AI"
)

demo.launch(share=True)
```
